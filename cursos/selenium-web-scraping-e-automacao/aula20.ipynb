```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from selenium.common import NoSuchElementException, TimeoutException, MoveTargetOutOfBoundsException
from selenium.webdriver.support.wait import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
# ActionChains
from selenium.webdriver.common.action_chains import ActionChains


chrome = webdriver.Chrome()
chrome.implicitly_wait(time_to_wait=10)  # aguardar 10 seg antes de levantar exceção

import time
import random
```

```python
# Exemplo 07
# https://curso-web-scraping.pages.dev/#/exemplo/7
# server local
pagina = 'http://localhost:8000/#/exemplo/7'
chrome.get(pagina)
```

```python
galeria = chrome.find_element(By.ID, 'galeria')
galeria
```

```python
imagens = galeria.find_elements(By.CSS_SELECTOR, 'img')
print(f'Imagens:{len(imagens)}')
for imagem in imagens:
    
    print('\t-', imagem.get_attribute('id'))
```

ActionChains(driver)  
- .metodos() - ações  
- .perform() - execução  

```python
#movever para elemento
time.sleep(2)
ActionChains(chrome).move_to_element(imagens[0]).perform()
time.sleep(2)
ActionChains(chrome).move_to_element(imagens[-1]).perform()
```

```python
for i in imagens:
    # time.sleep(1)
    ActionChains(chrome).move_to_element(i).perform()
    print(i.get_attribute('src'))
```

```python
for i in range(len(imagens)):
    # time.sleep(1)
    img = random.choice(imagens)
    ActionChains(chrome).move_to_element(img).perform()
    # clientHeight
    # clientWidth
    coordenada = f"x:{img.get_property('x')} - y:{img.get_property('y')} - largura:{img.get_property('clientWidth')} - altura:{img.get_property('clientHeight')}."
    print(i, '-', coordenada)
```

offset (deslocamento)

```python
img = imagens[1]  # no meio
ActionChains(chrome).move_to_element(img).perform()
coordenada = f"x:{img.get_property('x')} - y:{img.get_property('y')} - largura:{img.get_property('clientWidth')} - altura:{img.get_property('clientHeight')}."
print(i, '-', coordenada)
movx = 1 # -1, 0 ou 1
movy = 3 # linhas visiveis abaixo
coord_x = (img.get_property('clientWidth') + 10)  * movx 
coord_y = (img.get_property('clientHeight') + 10) * movy
```

```python
try:
    ActionChains(chrome).move_by_offset(xoffset=coord_x, yoffset=coord_y).perform()
except MoveTargetOutOfBoundsException as e:
    print(f'x:{coord_x} e y:{coord_y} fora do campo visual.')
```
